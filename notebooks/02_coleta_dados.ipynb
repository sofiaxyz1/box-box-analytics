## Coleta de dados — temporada 2025

**Objetivo:** a função `processar_quali(ano, gp)` recebe o ano e o GP (nome ou número da rodada) e devolve uma tabela com uma linha por piloto, contendo equipe, tempos de setor (S1, S2, S3), tempo de volta e posição de grid.

**Como funciona:**
- Carrega a sessão de quali daquele GP e pega a melhor volta cronometrada de cada piloto
- Junta essa tabela com `session.results` pela abreviação do piloto (nunca pela ordem das linhas — a ordem pode não significar nada)
- Converte os tempos, que vêm como `Timedelta`, para segundos com `.total_seconds()`

**Bugs encontrados e corrigidos:**
1. **Piloto sem volta rápida:** `pick_fastest()` devolve `None` quando o piloto não tem nenhuma volta cronometrada válida na sessão (ex: Bearman bateu no treino livre da Austrália 2025 e não rodou no quali). Sem checar isso, a linha seguinte (`volta['Driver']`) quebra com `'NoneType' object is not subscriptable`. A correção pula esse piloto com `if volta is None: continue`.

**Loop pela temporada:** percorri as 24 corridas de 2025 (filtrando a Rodada 0, que é a pré-temporada e não tem quali) com `try/except`, pra garantir que uma falha numa corrida não derrubasse o processamento das outras.

**Resultado:** 475 linhas de um teórico máximo de 480 (24 corridas × 20 pilotos) — a diferença de 5 vem de pilotos sem volta cronometrada em alguma sessão ao longo do ano.

**Onde ficou salvo:** `data/processed/quali_2025.csv`
